Converting webp to jpg


```python
import os
from PIL import Image

def convert_all_to_jpg(root_dir, allowed_exts=("jpg","jpeg","png","webp"), target_ext="jpg"):
    """
    Walk through root_dir (and subfolders), find files with extensions in allowed_exts,
    convert them to JPEG format, save with same name but .jpg extension, and optionally
    remove the original file if desired.
    """
    for subdir, dirs, files in os.walk(root_dir):
        for fname in files:
            name, ext = os.path.splitext(fname)
            ext_lower = ext.lower().lstrip('.')
            if ext_lower in allowed_exts and ext_lower != target_ext:
                in_path   = os.path.join(subdir, fname)
                out_fname = f"{name}.{target_ext}"
                out_path  = os.path.join(subdir, out_fname)
                try:
                    img = Image.open(in_path).convert("RGB")  # Convert to RGB to avoid issues with transparency
                    img.save(out_path, format="JPEG", quality=95)
                    # Optionally: os.remove(in_path)
                    print(f"Converted {in_path} → {out_path}")
                except Exception as e:
                    print(f"Failed converting {in_path}: {e}")

# Example usage:
dataset_root = "/content/dataset"  # change to your dataset root path
convert_all_to_jpg(dataset_root)
```

Transfer tranning

```python
base_model = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3)
)

x = base_model.output
x = GlobalAveragePooling2D()(x)
predictions = Dense(len(CLASS_NAMES), activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=predictions)

for layer in base_model.layers:
    layer.trainable = False

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## Call backs

```python
checkpoint = ModelCheckpoint(
    'best_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max'
)
early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=3,
    mode='max',
    restore_best_weights=True
)
```

```python
import os, random, shutil

# specify fraction for validation
val_frac = 0.2

for cls in CLASS_NAMES:
    train_cls_dir = os.path.join(train_dir, cls)
    val_cls_dir   = os.path.join(val_dir,   cls)
    os.makedirs(val_cls_dir, exist_ok=True)
    all_imgs = os.listdir(train_cls_dir)
    random.shuffle(all_imgs)
    num_val = int(len(all_imgs) * val_frac)
    val_imgs = all_imgs[:num_val]
    for fname in val_imgs:
        src = os.path.join(train_cls_dir, fname)
        dst = os.path.join(val_cls_dir,   fname)
        shutil.move(src, dst)
```

## train initial phase

```python
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint, early_stop]
)
```

```python
def predict_image(img_path, model, class_indices, image_size=IMAGE_SIZE):
    img = image.load_img(img_path, target_size=image_size)
    x   = image.img_to_array(img)
    x   = np.expand_dims(x, axis=0)
    x   = x / 255.0
    preds = model.predict(x)
    class_idx = np.argmax(preds, axis=1)[0]
    label_map = {v: k for k, v in class_indices.items()}
    predicted_label = label_map[class_idx]
    confidence = preds[0][class_idx]
    return predicted_label, confidence
```
